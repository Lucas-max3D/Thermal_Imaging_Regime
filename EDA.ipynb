```python
import os
import numpy as np
import tifffile
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

def show_thermal_video(cube_number, condition_number, line_number,
                       data_dir="raw_data",
                       start_frame=0, end_frame=100):
    """
    Load and display an animation of the thermal video corresponding to the
    specified cube_number, condition_number, and line_number.

    Parameters
    ----------
    cube_number : int
        Which cube (1–4).
    condition_number : int
        Which processing condition (1–12).
    line_number : int
        Which line (1–5).
    data_dir : str
        Base directory where 'cubeX' folders reside.
    start_frame : int
        First frame index to display (0-based).
    end_frame : int
        Last frame index to display (exclusive). Use a smaller range to avoid
        reading huge files entirely into memory.
    """
    # Construct the path:
    #  - Folder:  e.g. "cube1"
    #  - Subfolder: e.g. "cube1_C001H001S0001"
    #  - TIF file: e.g. "cube1_C001H001S0001-00.tif"
    cube_folder = f"cube{cube_number}"
    condition_subfolder = f"cube{cube_number}_C001H001S00{condition_number:02d}"
    tif_filename = f"cube{cube_number}_C001H001S00{condition_number:02d}-{(line_number-1):02d}.tif"
    
    file_path = os.path.join(data_dir, cube_folder, condition_subfolder, tif_filename)
    
    print("Loading TIF file:", file_path)

    # ---- Reading the data ----
    # If the file is extremely large, consider reading frames in a loop
    # or limiting the number of frames you load at once.
    with tifffile.TiffFile(file_path) as tif:
        n_frames = len(tif.pages)
        print(f"Total frames in TIF: {n_frames}")
        
        # Guard to avoid going beyond the file's total frame count
        end_frame = min(end_frame, n_frames)
        
        # Read only the selected frames [start_frame : end_frame]
        frames = []
        for i in range(start_frame, end_frame):
            frame = tif.pages[i].asarray()
            frames.append(frame)

    frames = np.array(frames, dtype=np.uint16)  # shape: (num_loaded_frames, height, width)
    print(f"Loaded frames shape: {frames.shape}")

    # ---- Rescaling from uint16 to uint8 for display ----
    # A simple way is to do a linear stretch between min and max
    min_val = frames.min()
    max_val = frames.max()
    print(f"Frame pixel range: [{min_val}, {max_val}]")
    if max_val > min_val:
        frames_8bit = (255 * (frames - min_val) / (max_val - min_val)).astype(np.uint8)
    else:
        # If all frames have the same value, just cast directly
        frames_8bit = frames.astype(np.uint8)

    # ---- Prepare animation ----
    fig, ax = plt.subplots()
    # Show the first frame
    im = ax.imshow(frames_8bit[0], cmap='gray', animated=True)
    ax.set_title(f"Cube {cube_number}, Condition {condition_number}, Line {line_number}\n"
                 f"Frames {start_frame}-{end_frame-1}")

    def update(frame_idx):
        im.set_array(frames_8bit[frame_idx])
        return [im]

    anim = FuncAnimation(
        fig, update, frames=len(frames_8bit),
        interval=50, blit=True  # interval in ms
    )

    plt.close(fig)  # Prevent double-plot in Jupyter
    return HTML(anim.to_jshtml())

```

```python
show_thermal_video(cube_number=1, condition_number=1, line_number=1, start_frame=850, end_frame=1250)
```
